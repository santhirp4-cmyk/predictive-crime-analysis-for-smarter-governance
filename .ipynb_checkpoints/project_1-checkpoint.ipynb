```python
#Finding entires that were mispelled
from fuzzywuzzy import fuzz
from fuzzywuzzy import process



# Function to find similar strings using fuzzywuzzy
def find_similar_strings(target, choices, threshold=80):
    result, score = process.extractOne(target, choices)
    if score >= threshold:
        return result
    else:
        return None

# Create a set of unique city names
unique_cities = grouped_df['City'].unique()

# Loop through each city and find similar strings
similar_strings = {}
for city in unique_cities:
    similar = find_similar_strings(city, unique_cities)
    if similar and similar != city:
        similar_strings[city] = similar

# Display similar strings
for original, similar in similar_strings.items():
    print(f"Original: {original}, Similar: {similar}")
```

```python
# Identify rows to drop
rows_to_drop = set()
for index, row in grouped_df.iterrows():
    city = row['City']
    similar = find_similar_strings(city, unique_cities)
    if similar and similar != city:
        rows_to_drop.add(index)

# Drop identified rows
df_cleaned = grouped_df.drop(index=rows_to_drop).reset_index(drop=True)

# Display the cleaned DataFrame
df_cleaned
```

```python
#IMPORT REQUIRED LIBRARIES
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
##%% STEP 2: LOAD CLEANED DATA
# Assuming you have already done the cleaning and grouping
df_ml = df_grouped.copy()  # use your grouped & cleaned DataFrame
df_ml.head()
```

```python
##%% STEP 3: CREATE TARGET VARIABLE
# Binary classification: High crime if Total > median, else Low
median_total = df_ml['Total'].median()
df_ml['Crime_Level'] = df_ml['Total'].apply(lambda x: 1 if x > median_total else 0)
# Encode 'City' as numeric
le = LabelEncoder()
df_ml['City_encoded'] = le.fit_transform(df_ml['City'])
X=df_ml[['City_encoded']]
Y=df_ml['Crime_Level']

df_ml.head()
```

```python

#STEP 5: Train and Evaluate Models

models = {
    'Logistic Regression': LogisticRegression(),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=3),
    'Naive Bayes': GaussianNB()
}

results = {}

for name, model in models.items():
    if name in ['Logistic Regression', 'KNN']:
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_test_scaled)
    else:
         model.fit(X_train, y_train)
         y_pred = model.predict(X_test)
    
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    
    print(f"--- {name} ---")
    print("Accuracy:", acc)
    print("Classification Report:\n", classification_report(y_test, y_pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
    print("\n")

```

```python
#STEP 6: Visualize Model Accuracy

plt.figure(figsize=(8,5))
sns.barplot(x=list(results.keys()), y=list(results.values()))
plt.title("Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.ylim(0,1)
plt.show()

```

```python


#STEP 7: Feature Importance (Tree-Based Models)

# Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
feat_importance_dt = pd.Series(dt_model.feature_importances_, index=X.columns)
feat_importance_dt.sort_values().plot(kind='barh', title="Decision Tree Feature Importance")
plt.show()

# Random Forest
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
feat_importance_rf =pd.Series(rf_model.feature_importances_, index=X.columns)
feat_importance_rf.sort_values().plot(kind='barh', color='green', title="Random Forest Feature Importance")
plt.show()

```

```python
#hyperparametre_tunning 
from sklearn.model_selection import GridSearchCV

log_params = {
    'C': [0.01, 0.1, 1, 10],
    'solver': ['liblinear']
}

grid_log = GridSearchCV(
    LogisticRegression(max_iter=1000),
    log_params,
    cv=5,
    scoring='accuracy'
)

grid_log.fit(X_train_scaled, y_train)
print("Best Logistic Regression Parameters:", grid_log.best_params_)
print("Best Logistic Regression Accuracy:", grid_log.best_score_)
```

```python
dt_params = {
    'max_depth': [None, 5, 10, 20],
    'min_samples_split': [2, 5, 10]
}

grid_dt = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    dt_params,
    cv=5,
    scoring='accuracy'
)

grid_dt.fit(X_train, y_train)

print("Best Decision Tree Parameters:", grid_dt.best_params_)
print("Best Decision Tree Accuracy:", grid_dt.best_score_)

```

```python
rf_params = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}

grid_rf = GridSearchCV(
    RandomForestClassifier(random_state=42),
    rf_params,
    cv=5,
    scoring='accuracy'
)

grid_rf.fit(X_train, y_train)

print("Best Random Forest Parameters:", grid_rf.best_params_)
print("Best Random Forest Accuracy:", grid_rf.best_score_)
```

```python
knn_params = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}

grid_knn = GridSearchCV(
    KNeighborsClassifier(),
    knn_params,
    cv=5,
    scoring='accuracy'
)

grid_knn.fit(X_train_scaled, y_train)

print("Best KNN Parameters:", grid_knn.best_params_)
print("Best KNN Accuracy:", grid_knn.best_score_)


```

```python
nb_params = {
    'var_smoothing': np.logspace(-9, -1, 10)
}

grid_nb = GridSearchCV(
    GaussianNB(),
    nb_params,
    cv=5,
    scoring='accuracy'
)

grid_nb.fit(X_train_scaled, y_train)

print("Best Naive Bayes Parameters:", grid_nb.best_params_)
print("Best Naive Bayes Accuracy:", grid_nb.best_score_)
```
