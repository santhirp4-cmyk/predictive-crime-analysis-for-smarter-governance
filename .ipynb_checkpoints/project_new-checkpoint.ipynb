```python
#Finding entires that were mispelled
from fuzzywuzzy import fuzz
from fuzzywuzzy import process



# Function to find similar strings using fuzzywuzzy
def find_similar_strings(target, choices, threshold=80):
    result, score = process.extractOne(target, choices)
    if score >= threshold:
        return result
    else:
        return None

# Create a set of unique city names
unique_cities = grouped_df['City'].unique()

# Loop through each city and find similar strings
similar_strings = {}
for city in unique_cities:
    similar = find_similar_strings(city, unique_cities)
    if similar and similar != city:
        similar_strings[city] = similar

# Display similar strings
for original, similar in similar_strings.items():
    print(f"Original: {original}, Similar: {similar}")
```

```python
# Identify rows to drop
rows_to_drop = set()
for index, row in grouped_df.iterrows():
    city = row['City']
    similar = find_similar_strings(city, unique_cities)
    if similar and similar != city:
        rows_to_drop.add(index)

# Drop identified rows
df_cleaned = grouped_df.drop(index=rows_to_drop).reset_index(drop=True)

# Display the cleaned DataFrame
df_cleaned
```

```python
#IMPORT REQUIRED LIBRARIES
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
##%% STEP 2: LOAD CLEANED DATA
# Assuming you have already done the cleaning and grouping
df_ml = df_grouped.copy()  # use your grouped & cleaned DataFrame
df_ml.head()
```

```python
df_ml = df_grouped.copy()

q1 = df_ml['Total'].quantile(0.33)
q2 = df_ml['Total'].quantile(0.66)

def crime_category(x):
    if x <= q1:
        return 0   # Low
    elif x <= q2:
        return 1   # Medium
    else:
        return 2   # High

df_ml['Crime_Level'] = df_ml['Total'].apply(crime_category)
crime_features = [
    'Personal Revenge', 'Anger', 'Fraud', 'Extortion',
    'Causing Disrepute', 'Prank', 'Sexual Exploitation'
]

X = df_ml[crime_features]
y = df_ml['Crime_Level']
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train.shape, X_test.shape)
```

```python
#model trainig and accurancy
results = {}

# Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_scaled, y_train)
y_pred = lr.predict(X_test_scaled)
results['Logistic Regression'] = accuracy_score(y_test, y_pred)

# Decision Tree
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
results['Decision Tree'] = accuracy_score(y_test, y_pred)

# Random Forest
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
results['Random Forest'] = accuracy_score(y_test, y_pred)

# KNN
knn = KNeighborsClassifier()
knn.fit(X_train_scaled, y_train)
y_pred = knn.predict(X_test_scaled)
results['KNN'] = accuracy_score(y_test, y_pred)

# Naive Bayes
nb = GaussianNB()
nb.fit(X_train_scaled, y_train)
y_pred = nb.predict(X_test_scaled)
results['Naive Bayes'] = accuracy_score(y_test, y_pred)

for model, acc in results.items():
    print(model, "Accuracy:", acc)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=list(results.keys()), y=list(results.values()))
plt.title("Baseline Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.ylim(0,1)
plt.xticks(rotation=45)
plt.show()
```

```python
#hyperparametre tunning
# Import the necessary libraries
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# Define the parameter grid
log_params = {
    'C': [0.01, 0.1, 1, 10],       # Regularization strength
    'solver': ['liblinear']         # Solver choice
}

# Set up GridSearchCV
grid_log = GridSearchCV(
    estimator=LogisticRegression(max_iter=1000),
    param_grid=log_params,
    cv=5,                          # 5-fold cross-validation
    scoring='accuracy'             # Metric to optimize
)

# Fit the model on training data
grid_log.fit(X_train_scaled, y_train)

# Print the best parameters and score
print("Best Parameters:", grid_log.best_params_)
print("Best Cross-Validation Accuracy:", grid_log.best_score_)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# Logistic Regression
log_params = {'C':[0.01,0.1,1,10], 'solver':['liblinear']}
grid_log = GridSearchCV(LogisticRegression(max_iter=1000), log_params, cv=5, scoring='accuracy')
grid_log.fit(X_train_scaled, y_train)

# Decision Tree
dt_params = {'max_depth':[None,5,10,20], 'min_samples_split':[2,5,10]}
grid_dt = GridSearchCV(DecisionTreeClassifier(), dt_params, cv=5, scoring='accuracy')
grid_dt.fit(X_train_scaled, y_train)

# Random Forest
rf_params = {'n_estimators':[50,100,200], 'max_depth':[None,5,10]}
grid_rf = GridSearchCV(RandomForestClassifier(), rf_params, cv=5, scoring='accuracy')
grid_rf.fit(X_train_scaled, y_train)

# KNN
knn_params = {'n_neighbors':[3,5,7,9], 'weights':['uniform','distance']}
grid_knn = GridSearchCV(KNeighborsClassifier(), knn_params, cv=5, scoring='accuracy')
grid_knn.fit(X_train_scaled, y_train)

# Naive Bayes (no hyperparameters to tune usually, but you can wrap it for consistency)
grid_nb = GridSearchCV(GaussianNB(), {}, cv=5, scoring='accuracy')
grid_nb.fit(X_train_scaled, y_train)

# Best estimators
best_log = grid_log.best_estimator_
best_dt  = grid_dt.best_estimator_
best_rf  = grid_rf.best_estimator_
best_knn = grid_knn.best_estimator_
best_nb  = grid_nb.best_estimator_

print("Best Logistic Regression:", best_log)
print("Best Decision Tree:", best_dt)
print("Best Random Forest:", best_rf)
print("Best KNN:", best_knn)
print("Best Naive Bayes:", best_nb)
```

```python
df_ml_preds = df_ml[['City','Total']].copy()

# Logistic Regression
df_ml_preds['Predicted_Logistic'] = best_log.predict(scaler.transform(X))

# Decision Tree
df_ml_preds['Predicted_DT'] = best_dt.predict(X)

# Random Forest
df_ml_preds['Predicted_RF'] = best_rf.predict(X)

# KNN
df_ml_preds['Predicted_KNN'] = best_knn.predict(scaler.transform(X))

# Naive Bayes
df_ml_preds['Predicted_NB'] = best_nb.predict(scaler.transform(X))
```

```python
label_map = {0:'Low', 1:'Medium', 2:'High'}

for col in ['Predicted_Logistic','Predicted_DT','Predicted_RF','Predicted_KNN','Predicted_NB']:
    df_ml_preds[col] = df_ml_preds[col].map(label_map)
```

```python
high_crime_rf = df_ml_preds[df_ml_preds['Predicted_RF']=='High'] \
                    .sort_values(by='Total', ascending=False)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(data=high_crime_rf, x='City', y='Total', palette='Reds_r')
plt.xticks(rotation=90)
plt.title("Predicted High Crime Cities - Random Forest")
plt.ylabel("Total Crimes")
plt.xlabel("City")
plt.show()
```
